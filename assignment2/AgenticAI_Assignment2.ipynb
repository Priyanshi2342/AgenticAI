```python
import os
from groq import Groq
from ddgs import DDGS
import wikipedia

# API KEY
client = Groq(api_key="YOUR_API")


# 🔧 TOOL 1: WEB SEARCH
def web_search(query):
    results = []
    with DDGS() as ddgs:
        for r in ddgs.text(query, max_results=5):
            results.append(r['body'])
    return "\n".join(results)


# 🔧 TOOL 2: WIKIPEDIA
def wiki_search(query):
    try:
        return wikipedia.summary(query, sentences=5)
    except:
        return "No result found"


# 🤖 AGENT (ReAct Style)
def research_agent(topic):

    print("🔍 Searching Web...")
    web_data = web_search(topic)

    print("📚 Fetching Wikipedia...")
    wiki_data = wiki_search(topic)

    combined = f"""
    Web Data:
    {web_data}

    Wikipedia Data:
    {wiki_data}
    """

    prompt = f"""
    You are an Autonomous Research Agent.

    Topic: {topic}

    Based on the data below, generate a structured report.

    {combined}

    Format:
    - Cover Page
    - Title
    - Introduction
    - Key Findings
    - Challenges
    - Future Scope
    - Conclusion
    """

    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[{"role": "user", "content": prompt}]
    )

    return response.choices[0].message.content


# ▶️ RUN
topic = "Impact of AI in Healthcare"

report = research_agent(topic)

print("\n===== FINAL REPORT =====\n")
print(report)
```
